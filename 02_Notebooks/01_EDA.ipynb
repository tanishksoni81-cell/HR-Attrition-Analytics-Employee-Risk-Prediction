```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv(r"C:\Users\win10\OneDrive\Desktop\Data-Projects\Portfolio Project\HR Attrition Analytics & Prediction\01_Data\01_Raw_Data\HR-Employee-Attrition.csv")
```

```python
print("The Null Values Are:\n", df.isnull().sum()/len(df)*100)
```

```python
# Attrition Distribution
df['Attrition'].value_counts()
sns.countplot(data=df, x='Attrition')
plt.show()
```

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(12,8))
sns.heatmap(corr)
plt.show()
```

```python
# Salary VS Attrition 
sns.boxplot(x='Attrition',
            y='MonthlyIncome',
            data=df
            )
plt.show()
```

```python
# Overtime VS Attrition
pd.crosstab(df['OverTime'],df['Attrition'])
sns.countplot(data=df, x='OverTime', hue='Attrition')
plt.show()
```

```python
# Satisfaction Analysis 
sns.boxplot(x='Attrition', y='JobSatisfaction', data = df)
plt.show()
```
