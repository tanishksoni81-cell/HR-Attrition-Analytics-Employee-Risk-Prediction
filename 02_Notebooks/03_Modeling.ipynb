```python
import pandas as pd

train = pd.read_csv(r'C:\Users\win10\OneDrive\Desktop\Data-Projects\Portfolio Project\HR Attrition Analytics & Prediction\01_Data\02_Processed_Data/train.csv')
test = pd.read_csv(r'C:\Users\win10\OneDrive\Desktop\Data-Projects\Portfolio Project\HR Attrition Analytics & Prediction\01_Data\02_Processed_Data/test.csv')
```

```python
# Split

X_train = train.drop('AttritionFlag', axis=1)

y_train = train['AttritionFlag']

X_test = test.drop('AttritionFlag', axis=1)

y_test = test['AttritionFlag']
```

```python
# Model 1 : Logistic Regression 
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
```

```python
from sklearn.metrics import (accuracy_score,classification_report,roc_auc_score)

def evaluate(model):
    pred = model.predict(X_test)
    prob = model.predict_proba(X_test)[:,1]
    print("Accuracy:", accuracy_score(y_test, pred))
    print(classification_report(y_test, pred))
    print("ROC_AUC:", roc_auc_score(y_test, prob))
```

```python
evaluate(lr)
```

```python
# Model 2 : Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=300, random_state=42)
rf.fit(X_train, y_train)
```

```python
evaluate(rf)
```

```python
# Model 3: XGBoost
from xgboost import XGBClassifier
xgb = XGBClassifier(random_state=42)
xgb.fit(X_train, y_train)
```

```python
evaluate(xgb)
```

```python
# Use best model 
importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance':rf.feature_importances_
})

importance.sort_values(by='Importance', ascending=False).head(20)
```
